```python
import pandas as pd
import re

# Lista de reemplazos y limpieza para normalizar personajes
personajes_limpios = []

for p in personajes:
    if not isinstance(p, str):
        continue
    # Limpieza básica de espacios y minúsculas
    p_clean = p.strip().lower()

    # Reemplazar saltos de línea y tabulaciones por espacios
    p_clean = re.sub(r'\s+', ' ', p_clean)

    # Reglas de normalización
    if p_clean in ['sancho', 'sancho panza panza', 'sancho respondio']:
        p_clean = 'sancho panza'
    elif 'don quijote' in p_clean or p_clean == 'quijote':
        p_clean = 'don quijote de la mancha'
    elif p_clean == 'don fernando':
        p_clean = 'fernando'
    elif p_clean in ['hidalgo', 'hidalgo d quijote de la mancha']:
        p_clean = 'don quijote de la mancha'

    # Filtrar ruidos obvios y palabras muy cortas
    stopwords_per = {'razon', 'leia', 'alli', 'traia', 'cayo', 'aqui', 'dices', 'sabes', 'veia', 'decis', 'habeis', 'cayo', 'traia', 'p', 'd'}
    if p_clean not in stopwords_per and len(p_clean) > 2:
        personajes_limpios.append(p_clean.title())

# Crear DataFrame y guardar de nuevo el Excel
df_mejorado = pd.DataFrame(personajes_limpios, columns=['Personaje'])
df_mejorado.to_excel('/content/personajes.xlsx', index=False)

print(f'Total de personajes procesados y normalizados: {len(df_mejorado)}')
display(df_mejorado.head(50).value_counts())
```

```python
# Generar un Network red con los embedding basados en PER (Normalizado)
import networkx as nx
import matplotlib.pyplot as plt
from collections import Counter

# Unificamos los nombres en la lista de personajes extraídos por spaCy
personajes_normalizados = [
    'sancho panza' if p.strip() == 'sancho' else p
    for p in personajes
]

# 1. Filtrar entidades inválidas o muy raras para simplificar el grafo
conteo_personajes = Counter(personajes_normalizados)

# Filtramos términos comunes que spaCy confunde y seleccionamos personajes con más de 1 aparición
stopwords_personajes = {'razon', 'leia', 'alli', 'traia', 'cayo', 'aqui', 'dices', 'hidalgo', 'sabes', 'veia'}
personajes_validos = [p for p, count in conteo_personajes.items() if p not in stopwords_personajes and count > 1]

# 2. Encontrar co-ocurrencias por párrafos en el texto corregido
parrafos = corpus.split('\n\n')
G = nx.Graph()

# Añadir nodos
for p in personajes_validos:
    G.add_node(p, size=conteo_personajes[p])

# Añadir conexiones (aristas)
for parrafo in parrafos:
    parrafo_low = parrafo.lower()
    # Identificar qué personajes válidos están presentes en este párrafo
    presentes = [p for p in personajes_validos if p in parrafo_low]
    if len(presentes) > 1:
        for i in range(len(presentes)):
            for j in range(i + 1, len(presentes)):
                p1, p2 = presentes[i], presentes[j]
                if G.has_edge(p1, p2):
                    G[p1][p2]['weight'] += 1
                else:
                    G.add_edge(p1, p2, weight=1)

# 3. Dibujar la red
plt.figure(figsize=(14, 10))
pos = nx.spring_layout(G, k=0.5, iterations=50)

# Filtrar nodos con pocas conexiones para un grafo más limpio
nodos_principales = [node for node, degree in dict(G.degree()).items() if degree > 2]
subG = G.subgraph(nodos_principales)
pos_sub = nx.spring_layout(subG, k=0.4, iterations=50)

# Tamaños de los nodos basados en su frecuencia
node_sizes = [conteo_personajes[node] * 30 for node in subG.nodes()]

# Dibujar elementos
nx.draw_networkx_nodes(subG, pos_sub, node_size=node_sizes, node_color='skyblue', alpha=0.8)
nx.draw_networkx_edges(subG, pos_sub, width=[d['weight']*0.5 for u, v, d in subG.edges(data=True)], edge_color='gray', alpha=0.5)
x = nx.draw_networkx_labels(subG, pos_sub, font_size=9, font_family='sans-serif')

plt.title("Red de Co-ocurrencia de Personajes en El Quijote (Normalizado)", fontsize=16)
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
import re

# 1. Partimos del texto limpio en minúsculas para realizar el reemplazo definitivo
quijote = quijote.lower()

# Corregimos preventivamente cualquier redundancia anterior ("sancho panza panza" -> "sancho panza")
quijote = re.sub(r"\bsancho\s+panza\s+panza\b", "sancho panza", quijote)

# Buscamos cualquier 'sancho' que NO esté seguido de 'panza' y lo convertimos a 'sancho panza'
patron_sancho_solo = r"\bsancho\b(?!\s+panza\b)"
quijote = re.sub(patron_sancho_solo, "sancho panza", quijote)

# También nos aseguramos de normalizar "don fernando" a "fernando"
quijote = re.sub(r"\bdon fernando\b", "fernando", quijote)

# 2. Verificación final sobre la variable principal
conteo_total = quijote.count("sancho panza")
sanchos_solos_restantes = len(re.findall(patron_sancho_solo, quijote))

print(f"--- Verificación de la variable 'quijote' ---")
print(f"- Frecuencia total de 'sancho panza': {conteo_total} veces.")
print(f"- Frecuencia de 'sancho' (aislado sin 'panza'): {sanchos_solos_restantes} veces.")
```

### Exportación de la Red para VOSviewer

VOSviewer es una herramienta excelente para visualizar mapas bibliométricos y redes complejas de forma interactiva. A continuación, exportaremos nuestro grafo de co-ocurrencia de personajes (`subG`) en dos formatos compatibles para que puedas abrirlo directamente en VOSviewer:
1. **Archivo GML (`red_quijote_vos.gml`)**: Un estándar de grafos muy fácil de leer.
2. **Archivos CSV de Nodos y Enlaces**: Ideales para la opción *'Create a map based on network data'* -> *'VOSviewer network file / GML file'* o importando las tablas de datos directamente.

```python
import networkx as nx
import pandas as pd

# 1. Exportar en formato GML (VOSviewer lo lee directamente en File -> Open)
nx.write_gml(subG, "red_quijote_vos.gml")
print("¡Archivo 'red_quijote_vos.gml' exportado correctamente!")

# 2. Alternativa: Crear y guardar archivos de texto/CSV para nodos y enlaces
# Tabla de nodos (id, label, weight/size)
nodos_data = []
for node, attrs in subG.nodes(data=True):
    nodos_data.append({
        "Id": node,
        "Label": node,
        "Weight": attrs.get('size', 1)
    })
df_nodos = pd.DataFrame(nodos_data)
df_nodos.to_csv("vos_nodos.csv", index=False)

# Tabla de aristas/enlaces (source, target, weight)
enlaces_data = []
for u, v, attrs in subG.edges(data=True):
    enlaces_data.append({
        "Source": u,
        "Target": v,
        "Weight": attrs.get('weight', 1)
    })
df_enlaces = pd.DataFrame(enlaces_data)
df_enlaces.to_csv("vos_enlaces.csv", index=False)

print("¡Archivos CSV 'vos_nodos.csv' y 'vos_enlaces.csv' exportados para VOSviewer!")
```
