```python
# Imputar todos los NaN con vecinos cercanos (K.means) con naigiboard=5 de dfMediterraneo
from sklearn.impute import KNNImputer

# Select only numerical columns for KNNImputer
numerical_cols = dfMediterraneo.select_dtypes(include=['number']).columns
dfMediterraneo_numerical = dfMediterraneo[numerical_cols]

imputer = KNNImputer(n_neighbors=5)
dfMediterraneo_imputed_array = imputer.fit_transform(dfMediterraneo_numerical)

# Convert the imputed array back to a DataFrame with column names
dfMediterraneo_imputed = pd.DataFrame(dfMediterraneo_imputed_array, columns=numerical_cols, index=dfMediterraneo_numerical.index)

# Optionally, you might want to merge this back with the non-numerical columns
for col in dfMediterraneo.select_dtypes(exclude=['number']).columns:
    dfMediterraneo_imputed[col] = dfMediterraneo[col]

# Reorder columns to match original if needed, or simply assign back to dfMediterraneo
dfMediterraneo = dfMediterraneo_imputed[dfMediterraneo.columns]
```

```python
# Test de Anova utilizando skelearn.features_selection import f_classif y me agregue un * o ** 0 +++ para la significancia
from sklearn.feature_selection import f_classif
import pandas as pd

# Separar el features (X: RainToday) y target (y)
X = dfMediterraneo.drop('RainToday', axis=1)
y = dfMediterraneo['RainToday']

# Select only numerical columns for X to be used with f_classif
X_numerical = X.select_dtypes(include=['number'])

# Aplicar el test de ANOVA
f_values, p_values = f_classif(X_numerical, y)

anova_df = pd.DataFrame({'Feature': X_numerical.columns, 'F-value': f_values, 'p-value': p_values})
anova_df = anova_df.sort_values(by='p-value')

# Add significance stars (optional, as requested in original comment idea)
def add_significance_stars(p_value):
    if p_value < 0.001:
        return '***'
    elif p_value < 0.01:
        return '**'
    elif p_value < 0.05:
        return '*'
    else:
        return ''

anova_df['Significance'] = anova_df['p-value'].apply(add_significance_stars)


display(anova_df)
```

```python
# Testear y Entrenar dfMediaterraneo a 50% y 50%
from sklearn.model_selection import train_test_split
X = dfMediterraneo.drop('RainToday', axis=1)
y = dfMediterraneo['RainToday']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
# Generar el modelo de regresion Logistica
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# Ingresar los datos de testeo
X_train = X_train[['Humidity3pm', 'Temp3pm', 'MaxTemp', 'Cloud9am', 'Cloud3pm', 'Temp9am', 'Pressure9am']]
X_test = X_test[['Humidity3pm', 'Temp3pm', 'MaxTemp', 'Cloud9am', 'Cloud3pm', 'Temp9am', 'Pressure9am']]
predic=LogisticRegression()
predic.fit(X_train, y_train)
y_pred = predic.predict(X_test)
```
